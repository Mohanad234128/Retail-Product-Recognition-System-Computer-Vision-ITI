```python
import tensorflow as tf
import os
from keras.utils import img_to_array , load_img
import numpy as np
import cv2
```

# load the model

```python
model = tf.keras.models.load_model('model1.h5')
print(model.summary())
```

# load the categories

```python
source_folder= r'C:\Self study\Machinie Learning\iTi\fruits-360_100x100\fruits-360\Test'
categories = os.listdir(source_folder)
categories.sort()
print(categories)
numofClasses = len(categories)
print(numofClasses)
```

# Load and prepare image

```python
def prepareImage(pathForImage):
    image = load_img(pathForImage, target_size=(100, 100))
    imgResult = img_to_array(image)
    print(imgResult.shape)
    imgResult = np.expand_dims(imgResult, axis=0)
    print(imgResult.shape)
    imgResult=imgResult / 255.
    return imgResult
```

```python
testImagePath = r'C:\Self study\Machinie Learning\iTi\Red_Apple.jpg'
imageForModel = prepareImage(testImagePath)

resultArray = model.predict(imageForModel,verbose = 1)
answer = np.argmax(resultArray,axis=1)
print(answer[0])
text= categories[answer[0]]
print('predicted image:'+text)
```

# show the image with text

```python
img= cv2.imread(testImagePath)
font = cv2.FONT_HERSHEY_COMPLEX
cv2.putText(img,text,(0,50),font,1,(209,19,77),2)
cv2.imshow('image',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
